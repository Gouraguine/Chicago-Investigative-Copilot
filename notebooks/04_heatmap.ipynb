```python
import pandas as pd
import numpy as np
import folium

from folium.plugins import HeatMap
```

```python
df = pd.read_csv("../data/clean_chicago_crimes.csv")

df.shape
```

```python
df[['latitude', 'longitude']].isnull().sum()

geo_df = df[['latitude', 'longitude']].dropna().copy()
geo_df = geo_df[
    geo_df['latitude'].between(-90, 90) &
    geo_df['longitude'].between(-180, 180)
]
center_lat = geo_df['latitude'].mean()
center_lon = geo_df['longitude'].mean()

print(center_lat, center_lon)
```

```python
crime_map = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=11,
    tiles="OpenStreetMap"
)
```

```python
heat_data = (
    geo_df
    .round({'latitude': 3, 'longitude': 3})
    .groupby(['latitude', 'longitude'])
    .size()
    .reset_index(name='count')
)
heat_points = heat_data[
    ['latitude', 'longitude', 'count']
].values.tolist()
heat_data.head()
```

```python
HeatMap(
    heat_points,
    radius=12,
    blur=15,
    max_zoom=13
).add_to(crime_map)
crime_map
```
