```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw_chicago_crimes.csv")

df.info()
```

```python
df["hour"] = df["date"].dt.hour
df["day_of_week"] = df["date"].dt.day_name()
df["month"] = df["date"].dt.month
df["month_name"] = df["date"].dt.month_name()
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

df["season"] = df["month"].apply(get_season)
df["season"].value_counts()
```

```python
df["year_month"] = df["date"].dt.to_period("M")
df[["date", "year_month"]].head()
```

```python
df_map = df.dropna(
    subset=["latitude", "longitude"]
).copy()

print("Main dataset:", df.shape)
print("Map dataset:", df_map.shape)

df["location_description"] = (
    df["location_description"].fillna("UNKNOWN")
)

df["location_description"].isnull().sum()

print(df.isnull().sum())
print("Map rows:", len(df_map))
```

```python
print("Ward:")
print(df["ward"].value_counts().sort_index())

print("\nCommunity Area:")
print(df["community_area"].value_counts().sort_index())
```

```python
df["ward"] = df["ward"].astype("Int64")
df["community_area"] = df["community_area"].astype("Int64")
df[["ward", "community_area"]].dtypes
text_columns = [
    "primary_type",
    "description",
    "location_description"
]

for col in text_columns:
    df[col] = df[col].str.strip().str.upper()
print("Crime types:", df["primary_type"].nunique())
print("Descriptions:", df["description"].nunique())
print("Locations:", df["location_description"].nunique())
```

```python
df.to_csv(
    "../data/clean_chicago_crimes.csv",
    index=False
)
```
